```python
dvp = df.groupby('Categoria')['Preco_Normal'].std()
print(dvp)

# DESVIO PADRÃO
# Lácteos e Instantâneos-y-Sopas (Presença de Outliers Críticos)
# Análise: O Desvio Padrão é superior à Média e à Mediana.
# Refinamento: Estas categorias apresentam altíssima dispersão e uma assimetria positiva severa. O fato de o desvio
# padrão superar a média indica que os dados não estão em torno de um centro comum, mas sim espalhados por valores 
# extremos. Existem outliers (produtos premium ou kits grandes) que distorcem a média, tornando a mediana o único 
# indicador confiável do preço típico nessas categorias.

# Comidas-Preparadas (Consistência e Concentração)
# Análise: Preços estáveis e desvio padrão baixo.
# Refinamento: Esta categoria apresenta a maior estabilidade de preços. O desvio padrão significativamente menor 
# que a média e a mediana indica uma baixa variabilidade. Os preços estão concentrados em uma faixa estreita, 
# sugerindo que o mercado para esses produtos é padronizado, sem grandes discrepâncias entre o item mais barato
# e o mais caro.

# Congelados (Distribuição de Cauda Longa)
# Análise: Valores espalhados e média influenciada por preços altos.
# Refinamento: O cenário indica uma distribuição heterogênea. Como o desvio padrão é quase idêntico à média, 
# a "média" torna-se uma métrica fraca para representar o produto comum. Há um "gap" ou um espalhamento constante 
# entre os preços, indicando que a categoria mistura produtos de valores muito distintos 
# (ex: vegetais congelados simples vs. pratos prontos complexos).

# Beleza-y-Cuidado-Personal (Alta Amplitude)
# Análise: Diferença grande nos preços.
# Refinamento: Apresenta uma amplitude de preços considerável. O desvio padrão elevado em relação à mediana confirma
# que a categoria é composta por nichos diferentes (ex: itens de higiene básica vs. cosméticos de luxo). A média é 
# puxada para cima, não sendo o valor mais frequente encontrado pelo consumidor.

# Frutas e Verduras (Equilíbrio e Normalidade)
# Análise: Preços "certos" e sem grandes fugas da curva.
# Refinamento: São as categorias que mais se aproximam de uma distribuição normal. Em Verduras, a proximidade entre 
# todas as métricas sugere que os preços são previsíveis e homogêneos. Em Frutas, o desvio padrão entre a média e a
# mediana indica uma variação natural (sazonalidade ou tipos de frutas), mas sem os desvios extremos vistos em Lácteos.
```

```python
df_lacteos = df.loc[df['Categoria'] == 'lacteos']
fig = px.box(
    df_lacteos, 
    x="Categoria", 
    y="Preco_Normal", 
    points="all",           
    title="Distribuição Interativa de Preços: Lácteos",
    labels={"Preco_Normal": "Preço Normal ($)"},
    color="Categoria",
    template="plotly_white"
)
fig.update_layout(
    yaxis_title="Preço Normal",
    xaxis_title="Categoria",
    showlegend=False
)
fig.show()
```

```python
media_descontos = df.groupby('Categoria')['Desconto'].mean().sort_values(ascending=False).reset_index()
fig = px.bar(
    media_descontos, 
    x='Categoria', 
    y='Desconto',
    color='Desconto',              
    color_continuous_scale='Viridis', 
    title='Média de Descontos por Categoria de Produto',
    labels={'Desconto': 'Média de Desconto (R$)', 'Categoria': 'Categoria'},
    template='plotly_white'        
)
fig.update_layout(
    xaxis_tickangle=-45,           
    showlegend=False,
    coloraxis_showscale=False      
)
fig.show()
```

```python
df_agrupado = df.groupby(['Categoria', 'Marca']).agg(
    Media_Desconto=('Desconto', 'mean'),
    Qtd_Produtos=('Desconto', 'count')
).reset_index()
fig = px.treemap(
    df_agrupado, 
    path=['Categoria', 'Marca'], 
    values='Qtd_Produtos',       
    color='Media_Desconto',      
    color_continuous_scale='RdYlGn',
    title='Mapa: Tamanho por Qtd de Produtos e Cor por Média de Desconto',
    labels={'Media_Desconto': 'Média Desconto (R$)', 'Qtd_Produtos': 'Nº de Produtos'}
)
fig.update_traces(textinfo="label+value")
fig.show()
```
