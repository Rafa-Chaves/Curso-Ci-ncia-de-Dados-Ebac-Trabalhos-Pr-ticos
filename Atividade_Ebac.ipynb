```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

espaco = '-' * 150

data = {
    'Nome': ['Alice', 'Joao', 'Charlie', 'David', 'Eva', 'Diego', 'Denize', 'Claudio'],
    'Idade': [25, 30, 35, 40, 45, 60, 22, 24],
    'Profissão': ['Engenheiro', 'Médico', 'Professor', 'Advogado', 'Médico','Engenheiro', 'Estudante','Estudante'],
    'Salário': ['4500', '8000', '5000', '10000', '12000','15000', '1200','1500'],
    'Limite_Credito': ['2500', '4000', '4000', '1000', '10000','2000', '500','250'],
    'Historico_Inadimplencia': ['0', '0', '0', '1', '0','1', '0','1'],
    'Estado_Civil': ['Casamento', 'Casamento', 'Solteiro', 'Solteiro', 'Casamento','Solteiro', 'Solteiro','Solteiro'],
    'Imovel_Proprio': ['0', '0', '0', '1', '1','1', '0','0']
}

df = pd.DataFrame(data)
df.head(5)



colunas_conversao = ['Salário', 'Limite_Credito', 'Imovel_Proprio']
df[colunas_conversao] = df[colunas_conversao].apply(pd.to_numeric)
df_grouped = df.groupby('Profissão')[['Salário', 'Limite_Credito']].mean()


ax = df_grouped.plot(kind='bar', figsize=(10,6), width=0.8, color=['#69d', 'red'])
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', label_type='edge', padding=3)
plt.title('Média de Salário e Crédito por Profissão')
plt.ylabel('Limite de Credito')
plt.xlabel('Profissões')
plt.legend(title='Métricas', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()


print('\n')
print(espaco)
print('\n')


plt.figure(figsize=(10, 4))
sns.stripplot(data=df, x='Limite_Credito', y='Nome', hue='Imovel_Proprio', palette='viridis', size=9)
sns.rugplot(data=df, x='Limite_Credito', hue='Imovel_Proprio', palette='viridis', height=0.05)
plt.legend(title='Imóvel', bbox_to_anchor=(1, 1), loc='upper left')
plt.title('Credito por Imóvel')
plt.xlabel('Limite de Credito')
plt.ylabel('Nomes')
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()


print('\n')
print(espaco)
print('\n')


plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Limite_Credito', hue='Historico_Inadimplencia', 
             bins=8, kde=True, palette='icefire', alpha=0.6)
sns.move_legend(plt.gca(), "upper left", bbox_to_anchor=(1, 1), title='Inadimplência')
plt.title('Distribuição de Limite de Crédito por Histórico de Inadimplência')
plt.xlabel('Limite de Crédito')
plt.ylabel('Historico Inadimplencia')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout() 
plt.show()

f
```
